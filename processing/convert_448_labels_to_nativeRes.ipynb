```python
import pandas as pd
import numpy as np
```

```python

def project_coordinates(df, original_width=448, original_height=448, target_width=6080, target_height=3420):
    """
    Project bounding box coordinates from original image size to target image size.
    
    Parameters:
    df: DataFrame with x1, y1, x2, y2 columns
    original_width, original_height: Original image dimensions (448x448)
    target_width, target_height: Target image dimensions (6080x3420)
    
    Returns:
    DataFrame with additional projected coordinate columns
    """
    
    # Create a copy of the dataframe to avoid modifying the original
    df_projected = df.copy()
    
    # Calculate scaling factors
    width_scale = target_width / original_width
    height_scale = target_height / original_height
    
    print(f"Width scaling factor: {width_scale:.4f}")
    print(f"Height scaling factor: {height_scale:.4f}")
    
    # Project coordinates
    df_projected['x1_projected'] = df['x1'] * width_scale
    df_projected['y1_projected'] = df['y1'] * height_scale
    df_projected['x2_projected'] = df['x2'] * width_scale
    df_projected['y2_projected'] = df['y2'] * height_scale
    
    # Round to integers for pixel coordinates
    df_projected['x1_projected'] = df_projected['x1_projected'].astype(int)
    df_projected['y1_projected'] = df_projected['y1_projected'].astype(int)
    df_projected['x2_projected'] = df_projected['x2_projected'].astype(int)
    df_projected['y2_projected'] = df_projected['y2_projected'].astype(int)
    
    # Ensure coordinates are within bounds
    df_projected['x1_projected'] = np.clip(df_projected['x1_projected'], 0, target_width - 1)
    df_projected['y1_projected'] = np.clip(df_projected['y1_projected'], 0, target_height - 1)
    df_projected['x2_projected'] = np.clip(df_projected['x2_projected'], 0, target_width - 1)
    df_projected['y2_projected'] = np.clip(df_projected['y2_projected'], 0, target_height - 1)
    
    return df_projected
```

```python
df = pd.read_csv("/Users/cmbreen/Documents/snow/alaska_dataset/all_images/labels.csv")

# Project coordinates
df_projected = project_coordinates(df)



```
