```python
import glob
import os
import pandas as pd
```

```python
files = glob.glob('/Volumes/My Book/alaska/native_res/**/*', recursive=True)

# Filter out directories, keep only files
files = [f for f in files if os.path.isfile(f)]

# Filter out hidden files
files = [f for f in files if not os.path.basename(f).startswith('.')]

# Create DataFrame
df = pd.DataFrame({'filepath': files})

# Add additional columns for analysis
df['filename'] = df['filepath'].apply(os.path.basename)
df['directory'] = df['filepath'].apply(os.path.dirname)
df['extension'] = df['filepath'].apply(lambda x: os.path.splitext(x)[1])

# Export to CSV
df.to_csv('alaska_files.csv', index=False)
print(f"Exported {len(df)} files to alaska_files.csv")
```
